## 4. Model Inference Test

Loading the trained model and running a prediction on a sample image.

```python
detector = GestureDetector(model_path=MODEL_PATH)
print(f"Classes: {detector.classes}")
```

```python
# Pick a random image from the raw dataset
import random
all_images = list(RAW_DATA_DIR.glob('**/*.png')) + list(RAW_DATA_DIR.glob('**/*.jpg'))

if all_images:
    img_path = random.choice(all_images)
    img = Image.open(img_path)
    
    # Display image
    plt.imshow(img)
    plt.axis('off')
    plt.show()
    
    # Run prediction (Simulating a frame input)
    # Note: Detector expects a frame (numpy array) and an ROI.
    # Here we manipulate it slightly for the notebook test.
    import cv2
    img_np = np.array(img)
    # Mock ROI as full image for this test
    h, w, _ = img_np.shape
    # Convert RGB back to BGR since cv2/detector expects BGR from webcam usually
    img_bgr = cv2.cvtColor(img_np, cv2.COLOR_RGB2BGR)
    
    label, conf = detector.predict(img_bgr, (0, 0, w, h))
    print(f"Prediction: {label} ({conf:.4f})")
else:
    print("No images found to test.")
```
